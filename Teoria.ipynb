## Zmienne

Zmienne to kontenery do przechowywania wartości danych. W Pythonie nie trzeba deklarować typu zmiennej. Zmienna jest tworzona w momencie, gdy po raz pierwszy przypiszesz do niej wartość.

### Typy danych
Python ma wiele wbudowanych typów danych. Oto kilka najważniejszych:
- **str**: (string) - do przechowywania tekstu. Tekst umieszczamy w cudzysłowach `"` lub apostrofach `'`.
- **int**: (integer) - do przechowywania liczb całkowitych.
- **float**: - do przechowywania liczb zmiennoprzecinkowych (z częścią ułamkową).
- **bool**: (boolean) - do przechowywania wartości logicznych, `True` lub `False`.
- **list**: - do przechowywania wielu elementów w jednej zmiennej. Listy są uporządkowane i modyfikowalne. Elementy umieszczamy w nawiasach kwadratowych `[]`.
- **dict**: (dictionary) - do przechowywania danych w parach klucz-wartość. Słowniki są nieuporządkowane, modyfikowalne i indeksowane. Elementy umieszczamy w nawiasach klamrowych `{}`.

```python
tekst = "Witaj, świecie!" # str
liczba_calkowita = 21 # int
liczba_zmiennoprzecinkowa = 3.14 # float
prawda = True # bool
lista_owocow = ["jabłko", "banan", "wiśnia"] # list
slownik_osob = {"imię": "Jan", "wiek": 30} # dict

print(f'Typ zmiennej tekst: {type(tekst)}')
print(f'Typ zmiennej liczba_calkowita: {type(liczba_calkowita)}')
print(f'Typ zmiennej liczba_zmiennoprzecinkowa: {type(liczba_zmiennoprzecinkowa)}')
print(f'Typ zmiennej prawda: {type(prawda)}')
print(f'Typ zmiennej lista_owocow: {type(lista_owocow)}')
print(f'Typ zmiennej slownik_osob: {type(slownik_osob)}')
```

## Instrukcje warunkowe

Instrukcje warunkowe pozwalają na wykonanie różnych bloków kodu w zależności od tego, czy określony warunek jest prawdziwy (`True`) czy fałszywy (`False`). Używamy do tego słów kluczowych `if`, `elif` (else if) oraz `else`.

```python
a = 200
b = 33

# Sprawdza, czy b jest większe od a
if b > a:
  print("b jest większe od a")
# Jeśli pierwszy warunek jest fałszywy, sprawdza, czy a jest równe b
elif a == b:
  print("a i b są równe")
# Jeśli żaden z powyższych warunków nie jest prawdziwy, wykonuje ten blok
else:
  print("a jest większe od b")
```

## Pętle

Pętle służą do wielokrotnego wykonywania bloku kodu. Python ma dwa główne rodzaje pętli: `while` i `for`.

### Pętla while
Pętla `while` wykonuje blok kodu tak długo, jak długo warunek jest prawdziwy.

```python
i = 1
while i < 6:
  print(i)
  i += 1 # Ważne: inkrementacja i, w przeciwnym razie pętla będzie nieskończona!
```

### Pętla for
Pętla `for` służy do iterowania po sekwencji (takiej jak lista, krotka, słownik, zbiór lub ciąg znaków).

```python
# Iterowanie po liście
fruits = ["jabłko", "banan", "wiśnia"]
print("Iterowanie po liście:")
for x in fruits:
  print(x)
```

Pętli `for` można również używać z funkcją `range()`, aby wykonać blok kodu określoną liczbę razy.

```python
# Użycie range() do iteracji 5 razy (od 0 do 4)
print("
Użycie range():")
for i in range(5):
    print(f"Pętla wykonana {i+1} raz(y)")
```

## Funkcje

Funkcja to blok kodu, który jest zorganizowany, wielokrotnego użytku i służy do wykonania określonego zadania. Funkcje pomagają w podziale programu na mniejsze, zarządzalne części.

```python
# Definiowanie funkcji
def moja_funkcja():
  print("Witaj z wnętrza funkcji!")

# Wywołanie funkcji
moja_funkcja()
```

### Parametry i Argumenty
Informacje można przekazywać do funkcji jako argumenty. Argumenty są określane po nazwie funkcji, w nawiasach. Można dodać dowolną liczbę argumentów, oddzielając je przecinkami.

```python
# Funkcja z jednym parametrem 'fname'
def przywitaj_sie(fname):
  print(f"Cześć, {fname}!")

# Wywołanie funkcji z różnymi argumentami
przywitaj_sie("Emil")
przywitaj_sie("Tobiasz")
przywitaj_sie("Linus")
```

### Wartość zwrotna
Aby funkcja zwróciła wartość, użyj instrukcji `return`.

```python
# Funkcja, która przyjmuje liczbę jako parametr i zwraca jej pięciokrotność
def pomnoz_przez_piec(x):
  return 5 * x

wynik1 = pomnoz_przez_piec(3)
wynik2 = pomnoz_przez_piec(5)

print(f"Wynik dla 3: {wynik1}")
print(f"Wynik dla 5: {wynik2}")
print(f"Wynik dla 9: {pomnoz_przez_piec(9)}")
```
